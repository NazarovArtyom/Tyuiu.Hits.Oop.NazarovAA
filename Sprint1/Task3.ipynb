Создать базовый класс Ticket в C#, который будет представлять билеты на
различные мероприятия. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.


#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System.IO;

class Ticket {
    private int ticketId;
    private int rowNumber;
    private int seatNumber;
    private bool isReserved;

    public int TicketId {get {return ticketId;} set {ticketId = value;}}
    public int RowNumber {get {return rowNumber;} set {rowNumber = value;}}
    public int SeatNumber {get {return seatNumber;} set{seatNumber = value;}}

    public bool IsReserved {get {return isReserved;} set {isReserved = value;}}

    public Ticket(int id, int row, int seat) {
        TicketId = id;
        RowNumber = row;
        SeatNumber = seat;
    }

    public virtual void PrintTicketInfo() {
        Console.WriteLine($"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber}");
    }

    public virtual bool ReserveSeat() {
        if (IsReserved) {
            Console.WriteLine("Место уже зарезервировано");
            return false;
        }
        else {
            IsReserved = true;
            Console.WriteLine("Вы успешно зарезервировали место");
            return true;
        }
    }

    public virtual string GetTicketDetails() {
        return $"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber} Место зарезервировано? {(IsReserved ? "Да" : "Нет")}";
    }

    public bool EqualRowNumber(Ticket ticket) {
        return ticket.RowNumber == RowNumber;
    }
}

class ConcertTicket : Ticket {
    private string artistName;

    public string ArtistName {get {return artistName;} set {artistName = value;}}

    public ConcertTicket(int id, int row, int seat, string artistName) : base(id, row, seat) {
        ArtistName = artistName;
    }

    public override void PrintTicketInfo() {
        Console.WriteLine($"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber} Исполнитель: {ArtistName}");
    }
}

class PlayTicket : Ticket {
    private string playTitle;

    public string PlayTitle {get {return playTitle;} set {playTitle = value;}}

    public PlayTicket(int id, int row, int seat, string playTitle) : base(id, row, seat) {
        PlayTitle = playTitle;
    }

    public override string GetTicketDetails() {
        return $"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber} Название спектакля: {PlayTitle} Место зарезервировано? {(IsReserved ? "Да" : "Нет")}";
    }
}

class ConferenceTicket : Ticket {
    private string conferenceName;
    private int reservedSeat;

    public string ConferenceName {get {return conferenceName;} set {conferenceName = value;}}
    public int ReservedSeat {get {return reservedSeat;} set {reservedSeat = value;}}

    public ConferenceTicket(int id, int row, int seat, string conferenceName) : base(id, row, seat) {
        ConferenceName = conferenceName;
    }

    public override bool ReserveSeat() {
        if (IsReserved) {
            Console.WriteLine("Место уже зарезервировано");
            return false;
        }
        else {
            int guessSeat = Convert.ToInt32(Console.ReadLine());

            IsReserved = true;
            ReservedSeat = guessSeat;
            Console.WriteLine($"Место {guessSeat} успешно зарезервировано");
            return true;
        }
    }

    public bool EqualSeat(ConferenceTicket ticket) {
        return ticket.ReservedSeat == ReservedSeat;
    }
}

ConcertTicket concert = new ConcertTicket(1, 10, 5, "Рок-группа");
PlayTicket play = new PlayTicket(2, 5, 12, "Гамлет");

ConferenceTicket conferenceIT = new ConferenceTicket(3, 2, 12, "IT-Конференция");
ConferenceTicket conferenceEngener = new ConferenceTicket(4, 4, 12, "Инженерная конференция");

concert.PrintTicketInfo();
Console.WriteLine(concert.GetTicketDetails());
concert.ReserveSeat();
Console.WriteLine(concert.GetTicketDetails());

Console.WriteLine();

play.PrintTicketInfo();
Console.WriteLine(play.GetTicketDetails());
play.ReserveSeat();
Console.WriteLine(play.GetTicketDetails());

Console.WriteLine();

conferenceIT.PrintTicketInfo();
Console.WriteLine(conferenceIT.GetTicketDetails());
Console.WriteLine("Введите желаемое место для конференции:");

Console.SetIn(new StringReader("12"));
conferenceIT.ReserveSeat();

Console.WriteLine();

conferenceIT.PrintTicketInfo();

Console.WriteLine(conferenceEngener.GetTicketDetails());
Console.WriteLine("Введите желаемое место для конференции:");

Console.SetIn(new StringReader("12"));
conferenceEngener.ReserveSeat();

Console.WriteLine();

Console.WriteLine($"Места в конференциях {conferenceIT.ConferenceName} и {conferenceEngener.ConferenceName} {(conferenceIT.EqualSeat(conferenceEngener) ? "совпадают" : "не совпадают")}");
```
