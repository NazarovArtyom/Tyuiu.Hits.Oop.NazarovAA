```
using System;
using System.IO;
using System.Collections.Generic;

abstract class TicketBase {
    private int ticketId;
    private int rowNumber;
    private int seatNumber;
    private bool isReserved;
    private string ticketType;
    private string venueSection;
    private int gateNumber;
    private decimal serviceFee;
    private DateTime createdAt;
    private string holderName;
    private string contactEmail;
    private bool isVip;
    private int accessLevel;

    public int TicketId {get {return ticketId;} set {ticketId = value;}}
    public int RowNumber {get {return rowNumber;} set {rowNumber = value;}}
    public int SeatNumber {get {return seatNumber;} set{seatNumber = value;}}
    public bool IsReserved {get {return isReserved;} set {isReserved = value;}}
    public string TicketType {get {return ticketType;} set {ticketType = value;}}
    public string VenueSection {get {return venueSection;} set {venueSection = value;}}
    public int GateNumber {get {return gateNumber;} set {gateNumber = value;}}
    public decimal ServiceFee {get {return serviceFee;} set {serviceFee = value;}}
    public DateTime CreatedAt {get {return createdAt;} set {createdAt = value;}}
    public string HolderName {get {return holderName;} set {holderName = value;}}
    public string ContactEmail {get {return contactEmail;} set {contactEmail = value;}}
    public bool IsVip {get {return isVip;} set {isVip = value;}}
    public int AccessLevel {get {return accessLevel;} set {accessLevel = value;}}

    public delegate void TicketEventHandler(string message);
    public event TicketEventHandler TicketReserved;
    public event TicketEventHandler TicketUpdated;

    protected void RaiseTicketReserved(string message) {
        TicketReserved?.Invoke(message);
    }

    protected void RaiseTicketUpdated(string message) {
        TicketUpdated?.Invoke(message);
    }

    public TicketBase(int id, int row, int seat) {
        TicketId = id;
        RowNumber = row;
        SeatNumber = seat;
        CreatedAt = DateTime.Now;
        ServiceFee = 0;
        GateNumber = 1;
        VenueSection = "A";
        HolderName = "";
        ContactEmail = "";
        IsVip = false;
        AccessLevel = 0;
    }

    public abstract void PrintTicketInfo();

    public virtual bool ReserveSeat() {
        if (IsReserved) {
            Console.WriteLine("Место уже зарезервировано");
            return false;
        }
        else {
            IsReserved = true;
            Console.WriteLine("Вы успешно зарезервировали место");
            RaiseTicketReserved($"Билет {TicketId} зарезервирован");
            return true;
        }
    }

    public virtual bool ReserveSeat(int customSeat) {
        if (IsReserved) {
            Console.WriteLine("Место уже зарезервировано");
            return false;
        }
        else {
            SeatNumber = customSeat;
            IsReserved = true;
            Console.WriteLine($"Место {customSeat} успешно зарезервировано");
            RaiseTicketReserved($"Билет {TicketId} зарезервирован на место {customSeat}");
            return true;
        }
    }

    public virtual bool ReserveSeat(string promoCode) {
        if (IsReserved) {
            Console.WriteLine("Место уже зарезервировано");
            return false;
        }
        else {
            IsReserved = true;
            Console.WriteLine($"Место зарезервировано по промокоду {promoCode}");
            RaiseTicketReserved($"Билет {TicketId} зарезервирован по промокоду {promoCode}");
            return true;
        }
    }

    public virtual string GetTicketDetails() {
        return $"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber} Место зарезервировано? {(IsReserved ? "Да" : "Нет")}";
    }

    public bool EqualRowNumber(TicketBase ticket) {
        return ticket.RowNumber == RowNumber;
    }

    public virtual decimal CalculateTotal() {
        return 0;
    }

    public virtual string GetCategory() {
        return "Базовый билет";
    }

    public string GetVenueInfo() {
        return $"Сектор: {VenueSection}, Вход: {GateNumber}";
    }

    public bool IsNewTicket() {
        return (DateTime.Now - CreatedAt).TotalDays < 1;
    }

    public decimal GetTotalWithFee() {
        return CalculateTotal() + ServiceFee;
    }

    public void ApplyServiceFee(decimal fee) {
        ServiceFee = fee;
    }

    public void AssignHolder(string name, string email) {
        HolderName = name;
        ContactEmail = email;
        RaiseTicketUpdated($"Владелец билета {TicketId} изменён на {name}");
    }

    public string GetHolderInfo() {
        return string.IsNullOrEmpty(HolderName) ? "Владелец не указан" : $"{HolderName} ({ContactEmail})";
    }

    public void UpgradeToVip(int level) {
        IsVip = true;
        AccessLevel = level;
        RaiseTicketUpdated($"Билет {TicketId} повышен до VIP уровня {level}");
    }

    public bool IsAccessGranted(int requiredLevel) {
        return AccessLevel >= requiredLevel;
    }
}

interface IDiscountable {
    decimal GetDiscount();
}

interface IReportable {
    string GenerateReport();
}

class Ticket : TicketBase {
    private string eventName;
    private decimal price;
    private string hallName;
    private DateTime eventDate;
    private string organizer;
    private string ageRestriction;
    private bool hasParking;
    private int availableSeats;
    private string dressCode;
    private bool isOnline;
    private string language;
    private int maxPerPerson;

    public string EventName {get {return eventName;} set {eventName = value;}}
    public decimal Price {get {return price;} set {price = value;}}
    public string HallName {get {return hallName;} set {hallName = value;}}
    public DateTime EventDate {get {return eventDate;} set {eventDate = value;}}
    public string Organizer {get {return organizer;} set {organizer = value;}}
    public string AgeRestriction {get {return ageRestriction;} set {ageRestriction = value;}}
    public bool HasParking {get {return hasParking;} set {hasParking = value;}}
    public int AvailableSeats {get {return availableSeats;} set {availableSeats = value;}}
    public string DressCode {get {return dressCode;} set {dressCode = value;}}
    public bool IsOnline {get {return isOnline;} set {isOnline = value;}}
    public string Language {get {return language;} set {language = value;}}
    public int MaxPerPerson {get {return maxPerPerson;} set {maxPerPerson = value;}}

    public Ticket(int id, int row, int seat) : base(id, row, seat) {
        AgeRestriction = "0+";
        HasParking = false;
        AvailableSeats = 0;
        DressCode = "Свободный";
        IsOnline = false;
        Language = "Русский";
        MaxPerPerson = 4;
    }

    public override void PrintTicketInfo() {
        Console.WriteLine($"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber}");
    }

    public void SetEventInfo(string name, decimal price, string hall, DateTime date) {
        EventName = name;
        Price = price;
        HallName = hall;
        EventDate = date;
    }

    public void SetEventInfo(string name, decimal price, string hall, DateTime date, string organizerName) {
        EventName = name;
        Price = price;
        HallName = hall;
        EventDate = date;
        Organizer = organizerName;
    }

    public string GetEventSummary() {
        return $"{EventName} в {HallName} {EventDate.ToShortDateString()}";
    }

    public decimal GetFinalPrice() {
        return Price;
    }

    public bool IsExpensive() {
        return Price > 1000;
    }

    public int DaysUntilEvent() {
        return (EventDate - DateTime.Now).Days;
    }

    public override decimal CalculateTotal() {
        return Price;
    }

    public override string GetCategory() {
        return "Обычный билет";
    }

    public string GetAgeInfo() {
        return $"Возрастное ограничение: {AgeRestriction}";
    }

    public bool IsSoldOut() {
        return AvailableSeats == 0;
    }

    public void UpdateAvailableSeats(int count) {
        AvailableSeats = count;
    }

    public string GetOrganizerInfo() {
        return string.IsNullOrEmpty(Organizer) ? "Организатор не указан" : $"Организатор: {Organizer}";
    }

    public string GetDressCodeInfo() {
        return $"Дресс-код: {DressCode}";
    }

    public void MakeOnline() {
        IsOnline = true;
    }

    public bool IsAvailableOnline() {
        return IsOnline;
    }

    public string GetLanguageInfo() {
        return $"Язык мероприятия: {Language}";
    }

    public bool CanBuyMore(int currentCount) {
        return currentCount < MaxPerPerson;
    }
}

class ConcertTicket : Ticket {
    private string artistName;
    private string genre;
    private int fanZoneCount;
    private bool hasMerch;
    private int soundCheckMinutes;
    private string stageType;
    private string tourName;
    private bool isSoldOutFlag;
    private int encoreCount;
    private string lightingType;

    public string ArtistName {get {return artistName;} set {artistName = value;}}
    public string Genre {get {return genre;} set {genre = value;}}
    public int FanZoneCount {get {return fanZoneCount;} set {fanZoneCount = value;}}
    public bool HasMerch {get {return hasMerch;} set {hasMerch = value;}}
    public int SoundCheckMinutes {get {return soundCheckMinutes;} set {soundCheckMinutes = value;}}
    public string StageType {get {return stageType;} set {stageType = value;}}
    public string TourName {get {return tourName;} set {tourName = value;}}
    public bool IsSoldOutFlag {get {return isSoldOutFlag;} set {isSoldOutFlag = value;}}
    public int EncoreCount {get {return encoreCount;} set {encoreCount = value;}}
    public string LightingType {get {return lightingType;} set {lightingType = value;}}

    public ConcertTicket(int id, int row, int seat, string artistName) : base(id, row, seat) {
        ArtistName = artistName;
        HasMerch = false;
        SoundCheckMinutes = 0;
        StageType = "Основная";
        TourName = "";
        IsSoldOutFlag = false;
        EncoreCount = 0;
        LightingType = "Стандартное";
    }

    public override void PrintTicketInfo() {
        Console.WriteLine($"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber} Исполнитель: {ArtistName}");
    }

    public string GetArtistGenre() {
        return $"{ArtistName} ({Genre})";
    }

    public bool IsRockConcert() {
        return Genre == "Рок";
    }

    public override decimal CalculateTotal() {
        decimal total = Price;
        if (FanZoneCount > 0) {
            total += FanZoneCount * 200;
        }
        return total;
    }

    public override string GetCategory() {
        return "Концерт";
    }

    public bool HasMerchandise() {
        return HasMerch;
    }

    public string GetStageInfo() {
        return $"Сцена: {StageType}, Саундчек: {SoundCheckMinutes} мин.";
    }

    public void EnableMerch() {
        HasMerch = true;
    }

    public string GetTourInfo() {
        return string.IsNullOrEmpty(TourName) ? "Тур не указан" : $"Тур: {TourName}";
    }

    public void AddEncore() {
        EncoreCount++;
    }

    public int GetEncoreCount() {
        return EncoreCount;
    }

    public string GetLightingInfo() {
        return $"Освещение: {LightingType}";
    }
}

class PlayTicket : Ticket {
    private string playTitle;
    private string director;
    private int durationMinutes;
    private int intermissionCount;
    private string theaterTroupe;
    private bool hasProgram;
    private string playGenre;
    private string playwright;
    private int actorCount;
    private bool isPremiere;
    private string setDesigner;

    public string PlayTitle {get {return playTitle;} set {playTitle = value;}}
    public string Director {get {return director;} set {director = value;}}
    public int DurationMinutes {get {return durationMinutes;} set {durationMinutes = value;}}
    public int IntermissionCount {get {return intermissionCount;} set {intermissionCount = value;}}
    public string TheaterTroupe {get {return theaterTroupe;} set {theaterTroupe = value;}}
    public bool HasProgram {get {return hasProgram;} set {hasProgram = value;}}
    public string PlayGenre {get {return playGenre;} set {playGenre = value;}}
    public string Playwright {get {return playwright;} set {playwright = value;}}
    public int ActorCount {get {return actorCount;} set {actorCount = value;}}
    public bool IsPremiere {get {return isPremiere;} set {isPremiere = value;}}
    public string SetDesigner {get {return setDesigner;} set {setDesigner = value;}}

    public PlayTicket(int id, int row, int seat, string playTitle) : base(id, row, seat) {
        PlayTitle = playTitle;
        TheaterTroupe = "";
        HasProgram = false;
        PlayGenre = "Драма";
        Playwright = "";
        ActorCount = 0;
        IsPremiere = false;
        SetDesigner = "";
    }

    public override string GetTicketDetails() {
        return $"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber} Название спектакля: {PlayTitle} Место зарезервировано? {(IsReserved ? "Да" : "Нет")}";
    }

    public string GetPlayInfo() {
        return $"{PlayTitle}, режиссёр: {Director}, длительность: {DurationMinutes} мин.";
    }

    public bool IsLongPlay() {
        return DurationMinutes > 120;
    }

    public override decimal CalculateTotal() {
        decimal total = Price;
        if (IntermissionCount > 0) {
            total += IntermissionCount * 50;
        }
        return total;
    }

    public override string GetCategory() {
        return "Спектакль";
    }

    public string GetTroupeInfo() {
        return $"Труппа: {TheaterTroupe}, Жанр: {PlayGenre}";
    }

    public bool HasProgramBook() {
        return HasProgram;
    }

    public void IncludeProgram() {
        HasProgram = true;
    }

    public string GetPlaywrightInfo() {
        return string.IsNullOrEmpty(Playwright) ? "Автор не указан" : $"Автор: {Playwright}";
    }

    public string GetCastInfo() {
        return $"Количество актёров: {ActorCount}";
    }

    public bool IsNewProduction() {
        return IsPremiere;
    }

    public string GetSetDesignerInfo() {
        return string.IsNullOrEmpty(SetDesigner) ? "Художник не указан" : $"Художник: {SetDesigner}";
    }
}

class ConferenceTicket : Ticket, IDiscountable, IReportable {
    private string conferenceName;
    private int reservedSeat;
    private string speakerName;
    private int participantCount;
    private string topic;
    private bool hasCertificate;
    private int coffeeBreaks;
    private string trackName;
    private bool isHybrid;
    private int sessionCount;
    private string sponsorName;

    public string ConferenceName {get {return conferenceName;} set {conferenceName = value;}}
    public int ReservedSeat {get {return reservedSeat;} set {reservedSeat = value;}}
    public string SpeakerName {get {return speakerName;} set {speakerName = value;}}
    public int ParticipantCount {get {return participantCount;} set {participantCount = value;}}
    public string Topic {get {return topic;} set {topic = value;}}
    public bool HasCertificate {get {return hasCertificate;} set {hasCertificate = value;}}
    public int CoffeeBreaks {get {return coffeeBreaks;} set {coffeeBreaks = value;}}
    public string TrackName {get {return trackName;} set {trackName = value;}}
    public bool IsHybrid {get {return isHybrid;} set {isHybrid = value;}}
    public int SessionCount {get {return sessionCount;} set {sessionCount = value;}}
    public string SponsorName {get {return sponsorName;} set {sponsorName = value;}}

    public ConferenceTicket(int id, int row, int seat, string conferenceName) : base(id, row, seat) {
        ConferenceName = conferenceName;
        Topic = "";
        HasCertificate = false;
        CoffeeBreaks = 0;
        TrackName = "";
        IsHybrid = false;
        SessionCount = 0;
        SponsorName = "";
    }

    public override bool ReserveSeat() {
        if (IsReserved) {
            Console.WriteLine("Место уже зарезервировано");
            return false;
        }
        else {
            int guessSeat = Convert.ToInt32(Console.ReadLine());
            IsReserved = true;
            ReservedSeat = guessSeat;
            Console.WriteLine($"Место {guessSeat} успешно зарезервировано");
            RaiseTicketReserved($"Билет {TicketId} зарезервирован на место {guessSeat}");
            return true;
        }
    }

    public override bool ReserveSeat(int customSeat) {
        if (IsReserved) {
            Console.WriteLine("Место уже зарезервировано");
            return false;
        }
        else {
            IsReserved = true;
            ReservedSeat = customSeat;
            Console.WriteLine($"Место {customSeat} успешно зарезервировано для конференции");
            RaiseTicketReserved($"Билет {TicketId} зарезервирован на место {customSeat}");
            return true;
        }
    }

    public bool EqualSeat(ConferenceTicket ticket) {
        return ticket.ReservedSeat == ReservedSeat;
    }

    decimal IDiscountable.GetDiscount() {
        return Price * 0.1m;
    }

    string IReportable.GenerateReport() {
        return $"Отчёт по конференции {ConferenceName}: спикер {SpeakerName}, цена {Price}, скидка {((IDiscountable)this).GetDiscount()}";
    }

    public bool HasSpeaker() {
        return !string.IsNullOrEmpty(SpeakerName);
    }

    public string GetSpeakerInfo() {
        return $"Спикер: {SpeakerName}";
    }

    public override decimal CalculateTotal() {
        decimal total = Price - ((IDiscountable)this).GetDiscount();
        if (ParticipantCount > 1) {
            total *= ParticipantCount;
        }
        return total;
    }

    public override string GetCategory() {
        return "Конференция";
    }

    public string GetTopicInfo() {
        return $"Тема: {Topic}";
    }

    public bool ProvidesCertificate() {
        return HasCertificate;
    }

    public int GetTotalBreakTime() {
        return CoffeeBreaks * 15;
    }

    public string GetTrackInfo() {
        return string.IsNullOrEmpty(TrackName) ? "Трек не указан" : $"Трек: {TrackName}";
    }

    public void MakeHybrid() {
        IsHybrid = true;
    }

    public bool IsHybridFormat() {
        return IsHybrid;
    }

    public string GetSessionInfo() {
        return $"Количество сессий: {SessionCount}";
    }

    public string GetSponsorInfo() {
        return string.IsNullOrEmpty(SponsorName) ? "Спонсор не указан" : $"Спонсор: {SponsorName}";
    }
}

class TicketCollection<T> where T : TicketBase {
    private List<T> _tickets = new List<T>();

    public void Add(T ticket) {
        _tickets.Add(ticket);
    }

    public void Remove(T ticket) {
        _tickets.Remove(ticket);
    }

    public void PrintAll() {
        foreach (var ticket in _tickets) {
            ticket.PrintTicketInfo();
        }
    }

    public decimal GetTotalPrice() {
        decimal total = 0;
        foreach (var ticket in _tickets) {
            total += ticket.CalculateTotal();
        }
        return total;
    }

    public int Count {
        get { return _tickets.Count; }
    }

    public List<T> Find(Func<T, bool> predicate) {
        return _tickets.FindAll(new Predicate<T>(predicate));
    }

    public void ForEach(Action<T> action) {
        _tickets.ForEach(action);
    }
}

class DiscountService {
    private IDiscountable _discountable;

    public DiscountService(IDiscountable discountable) {
        _discountable = discountable;
    }

    public decimal CalculateDiscount() {
        return _discountable.GetDiscount();
    }
}

class ReportService {
    private IReportable _reportable;

    public ReportService(IReportable reportable) {
        _reportable = reportable;
    }

    public string BuildReport() {
        return _reportable.GenerateReport();
    }
}

class TicketNotifier {
    public delegate void NotifyHandler(string message);
    public event NotifyHandler OnNotify;

    public void Notify(string message) {
        OnNotify?.Invoke(message);
    }
}

ConcertTicket concert = new ConcertTicket(1, 10, 5, "Рок-группа");
PlayTicket play = new PlayTicket(2, 5, 12, "Гамлет");

ConferenceTicket conferenceIT = new ConferenceTicket(3, 2, 12, "IT-Конференция");
ConferenceTicket conferenceEngener = new ConferenceTicket(4, 4, 12, "Инженерная конференция");

concert.TicketReserved += msg => Console.WriteLine($"[СОБЫТИЕ] {msg}");
play.TicketReserved += msg => Console.WriteLine($"[СОБЫТИЕ] {msg}");
conferenceIT.TicketUpdated += msg => Console.WriteLine($"[ОБНОВЛЕНИЕ] {msg}");

concert.Genre = "Рок";
concert.FanZoneCount = 2;
concert.TourName = "Мировой тур 2024";
concert.LightingType = "Лазерное";
concert.AddEncore();
concert.SetEventInfo("Рок-фестиваль", 1500, "Главный зал", DateTime.Now.AddDays(10), "ООО Музыка");
concert.AssignHolder("Иванов Иван", "ivanov@mail.com");
concert.UpgradeToVip(2);

play.Director = "Иванов";
play.DurationMinutes = 150;
play.IntermissionCount = 2;
play.Playwright = "Шекспир";
play.ActorCount = 15;
play.SetDesigner = "Петрова";
play.SetEventInfo("Гамлет", 800, "Малый зал", DateTime.Now.AddDays(5));

conferenceIT.SpeakerName = "Петров";
conferenceIT.ParticipantCount = 3;
conferenceIT.TrackName = "Backend";
conferenceIT.SessionCount = 8;
conferenceIT.SponsorName = "TechCorp";
conferenceIT.SetEventInfo("IT-Конференция", 2000, "Конференц-зал", DateTime.Now.AddDays(20));
conferenceIT.AssignHolder("Сидоров Сидор", "sidorov@mail.com");

conferenceEngener.SpeakerName = "Сидоров";
conferenceEngener.TrackName = "Frontend";
conferenceEngener.MakeHybrid();
conferenceEngener.SetEventInfo("Инженерная конференция", 1800, "Зал Б", DateTime.Now.AddDays(30));

concert.PrintTicketInfo();
Console.WriteLine(concert.GetTicketDetails());
Console.WriteLine(concert.GetArtistGenre());
Console.WriteLine(concert.IsRockConcert());
Console.WriteLine(concert.GetEventSummary());
Console.WriteLine(concert.IsExpensive());
Console.WriteLine(concert.DaysUntilEvent());
Console.WriteLine($"Категория: {concert.GetCategory()}, Итог: {concert.CalculateTotal()}");
concert.ReserveSeat(7);
Console.WriteLine(concert.GetTicketDetails());
Console.WriteLine(concert.GetHolderInfo());
Console.WriteLine($"VIP: {concert.IsVip}, Доступ: {concert.IsAccessGranted(2)}");
Console.WriteLine(concert.GetTourInfo());
Console.WriteLine($"Бисы: {concert.GetEncoreCount()}");
Console.WriteLine(concert.GetLightingInfo());

Console.WriteLine();

play.PrintTicketInfo();
Console.WriteLine(play.GetTicketDetails());
Console.WriteLine(play.GetPlayInfo());
Console.WriteLine(play.IsLongPlay());
Console.WriteLine(play.GetEventSummary());
Console.WriteLine($"Категория: {play.GetCategory()}, Итог: {play.CalculateTotal()}");
play.ReserveSeat("SUMMER2024");
Console.WriteLine(play.GetTicketDetails());
Console.WriteLine(play.GetPlaywrightInfo());
Console.WriteLine(play.GetCastInfo());
Console.WriteLine($"Премьера: {play.IsNewProduction()}");
Console.WriteLine(play.GetSetDesignerInfo());

Console.WriteLine();

conferenceIT.PrintTicketInfo();
Console.WriteLine(conferenceIT.GetTicketDetails());
Console.WriteLine(conferenceIT.HasSpeaker());
Console.WriteLine(conferenceIT.GetSpeakerInfo());
Console.WriteLine(((IReportable)conferenceIT).GenerateReport());
Console.WriteLine($"Категория: {conferenceIT.GetCategory()}, Итог: {conferenceIT.CalculateTotal()}");
Console.WriteLine(conferenceIT.GetTrackInfo());
Console.WriteLine(conferenceIT.GetSessionInfo());
Console.WriteLine(conferenceIT.GetSponsorInfo());

Console.SetIn(new StringReader("12"));
conferenceIT.ReserveSeat();

Console.WriteLine();

conferenceEngener.PrintTicketInfo();
Console.WriteLine(conferenceEngener.GetTicketDetails());
Console.WriteLine(((IReportable)conferenceEngener).GenerateReport());
Console.WriteLine($"Гибридный формат: {conferenceEngener.IsHybridFormat()}");

Console.WriteLine($"Места в конференциях {conferenceIT.ConferenceName} и {conferenceEngener.ConferenceName} {(conferenceIT.EqualSeat(conferenceEngener) ? "совпадают" : "не совпадают")}");

Console.WriteLine();

List<TicketBase> allTickets = new List<TicketBase> { concert, play, conferenceIT, conferenceEngener };
foreach (var t in allTickets) {
    Console.WriteLine($"{t.GetCategory()} -> {t.CalculateTotal()}");
}

Console.WriteLine();

TicketCollection<ConcertTicket> concerts = new TicketCollection<ConcertTicket>();
concerts.Add(concert);
concerts.Add(new ConcertTicket(5, 1, 1, "Джаз-бэнд"));
concerts.PrintAll();
Console.WriteLine($"Всего концертов: {concerts.Count}, Общая стоимость: {concerts.GetTotalPrice()}");

TicketCollection<TicketBase> mixed = new TicketCollection<TicketBase>();
mixed.Add(concert);
mixed.Add(play);
mixed.Add(conferenceIT);
mixed.PrintAll();
Console.WriteLine($"Всего билетов: {mixed.Count}, Общая стоимость: {mixed.GetTotalPrice()}");

var vipTickets = mixed.Find(t => t.IsVip);
Console.WriteLine($"VIP билетов: {vipTickets.Count}");
mixed.ForEach(t => Console.WriteLine($"Обработка билета {t.TicketId}: {t.GetCategory()}"));

Console.WriteLine();

DiscountService discountService = new DiscountService(conferenceIT);
Console.WriteLine($"Скидка через сервис: {discountService.CalculateDiscount()}");

ReportService reportService = new ReportService(conferenceIT);
Console.WriteLine(reportService.BuildReport());

Console.WriteLine(concert.GetVenueInfo());
Console.WriteLine(concert.GetStageInfo());
concert.EnableMerch();
Console.WriteLine($"Мерч доступен: {concert.HasMerchandise()}");

Console.WriteLine(play.GetTroupeInfo());
play.IncludeProgram();
Console.WriteLine($"Программа доступна: {play.HasProgramBook()}");

Console.WriteLine(conferenceIT.GetTopicInfo());
Console.WriteLine($"Сертификат: {conferenceIT.ProvidesCertificate()}, Время перерывов: {conferenceIT.GetTotalBreakTime()} мин.");

concert.ApplyServiceFee(150);
Console.WriteLine($"Итог с сервисным сбором: {concert.GetTotalWithFee()}");

Console.WriteLine();

TicketNotifier notifier = new TicketNotifier();
notifier.OnNotify += message => Console.WriteLine($"[УВЕДОМЛЕНИЕ] {message}");
notifier.Notify("Система билетов готова к работе");

Dictionary<int, TicketBase> ticketDictionary = new Dictionary<int, TicketBase>();
ticketDictionary.Add(concert.TicketId, concert);
ticketDictionary.Add(play.TicketId, play);
ticketDictionary.Add(conferenceIT.TicketId, conferenceIT);
ticketDictionary.Add(conferenceEngener.TicketId, conferenceEngener);

Console.WriteLine($"\nВсего билетов в словаре: {ticketDictionary.Count}");
foreach (var pair in ticketDictionary) {
    Console.WriteLine($"Ключ {pair.Key}: {pair.Value.GetCategory()} -> {pair.Value.CalculateTotal()}");
}

HashSet<string> uniqueCategories = new HashSet<string>();
foreach (var ticket in ticketDictionary.Values) {
    uniqueCategories.Add(ticket.GetCategory());
}
Console.WriteLine($"Уникальные категории: {string.Join(", ", uniqueCategories)}");
```
