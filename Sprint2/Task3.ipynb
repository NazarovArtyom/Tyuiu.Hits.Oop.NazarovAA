Создать базовый класс Ticket в C#, который будет представлять билеты на
различные мероприятия. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.IO;

abstract class TicketBase {
    private int ticketId;
    private int rowNumber;
    private int seatNumber;
    private bool isReserved;

    public int TicketId {get {return ticketId;} set {ticketId = value;}}
    public int RowNumber {get {return rowNumber;} set {rowNumber = value;}}
    public int SeatNumber {get {return seatNumber;} set{seatNumber = value;}}

    public bool IsReserved {get {return isReserved;} set {isReserved = value;}}

    public TicketBase(int id, int row, int seat) {
        TicketId = id;
        RowNumber = row;
        SeatNumber = seat;
    }

    public abstract void PrintTicketInfo();

    public virtual bool ReserveSeat() {
        if (IsReserved) {
            Console.WriteLine("Место уже зарезервировано");
            return false;
        }
        else {
            IsReserved = true;
            Console.WriteLine("Вы успешно зарезервировали место");
            return true;
        }
    }

    public virtual string GetTicketDetails() {
        return $"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber} Место зарезервировано? {(IsReserved ? "Да" : "Нет")}";
    }

    public bool EqualRowNumber(TicketBase ticket) {
        return ticket.RowNumber == RowNumber;
    }
}

interface IDiscountable {
    decimal GetDiscount();
}

interface IReportable {
    string GenerateReport();
}

class Ticket : TicketBase {
    private string eventName;
    private decimal price;
    private string hallName;
    private DateTime eventDate;

    public string EventName {get {return eventName;} set {eventName = value;}}
    public decimal Price {get {return price;} set {price = value;}}
    public string HallName {get {return hallName;} set {hallName = value;}}
    public DateTime EventDate {get {return eventDate;} set {eventDate = value;}}

    public Ticket(int id, int row, int seat) : base(id, row, seat) {
    }

    public override void PrintTicketInfo() {
        Console.WriteLine($"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber}");
    }

    public void SetEventInfo(string name, decimal price, string hall, DateTime date) {
        EventName = name;
        Price = price;
        HallName = hall;
        EventDate = date;
    }

    public string GetEventSummary() {
        return $"{EventName} в {HallName} {EventDate.ToShortDateString()}";
    }

    public decimal GetFinalPrice() {
        return Price;
    }

    public bool IsExpensive() {
        return Price > 1000;
    }

    public int DaysUntilEvent() {
        return (EventDate - DateTime.Now).Days;
    }
}

class ConcertTicket : Ticket {
    private string artistName;
    private string genre;

    public string ArtistName {get {return artistName;} set {artistName = value;}}
    public string Genre {get {return genre;} set {genre = value;}}

    public ConcertTicket(int id, int row, int seat, string artistName) : base(id, row, seat) {
        ArtistName = artistName;
    }

    public override void PrintTicketInfo() {
        Console.WriteLine($"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber} Исполнитель: {ArtistName}");
    }

    public string GetArtistGenre() {
        return $"{ArtistName} ({Genre})";
    }

    public bool IsRockConcert() {
        return Genre == "Рок";
    }
}

class PlayTicket : Ticket {
    private string playTitle;
    private string director;
    private int durationMinutes;

    public string PlayTitle {get {return playTitle;} set {playTitle = value;}}
    public string Director {get {return director;} set {director = value;}}
    public int DurationMinutes {get {return durationMinutes;} set {durationMinutes = value;}}

    public PlayTicket(int id, int row, int seat, string playTitle) : base(id, row, seat) {
        PlayTitle = playTitle;
    }

    public override string GetTicketDetails() {
        return $"ID: {TicketId} Ряд: {RowNumber} Место: {SeatNumber} Название спектакля: {PlayTitle} Место зарезервировано? {(IsReserved ? "Да" : "Нет")}";
    }

    public string GetPlayInfo() {
        return $"{PlayTitle}, режиссёр: {Director}, длительность: {DurationMinutes} мин.";
    }

    public bool IsLongPlay() {
        return DurationMinutes > 120;
    }
}

class ConferenceTicket : Ticket, IDiscountable, IReportable {
    private string conferenceName;
    private int reservedSeat;
    private string speakerName;

    public string ConferenceName {get {return conferenceName;} set {conferenceName = value;}}
    public int ReservedSeat {get {return reservedSeat;} set {reservedSeat = value;}}
    public string SpeakerName {get {return speakerName;} set {speakerName = value;}}

    public ConferenceTicket(int id, int row, int seat, string conferenceName) : base(id, row, seat) {
        ConferenceName = conferenceName;
    }

    public override bool ReserveSeat() {
        if (IsReserved) {
            Console.WriteLine("Место уже зарезервировано");
            return false;
        }
        else {
            int guessSeat = Convert.ToInt32(Console.ReadLine());

            IsReserved = true;
            ReservedSeat = guessSeat;
            Console.WriteLine($"Место {guessSeat} успешно зарезервировано");
            return true;
        }
    }

    public bool EqualSeat(ConferenceTicket ticket) {
        return ticket.ReservedSeat == ReservedSeat;
    }

    public decimal GetDiscount() {
        return Price * 0.1m;
    }

    public string GenerateReport() {
        return $"Отчёт по конференции {ConferenceName}: спикер {SpeakerName}, цена {Price}, скидка {GetDiscount()}";
    }

    public bool HasSpeaker() {
        return !string.IsNullOrEmpty(SpeakerName);
    }

    public string GetSpeakerInfo() {
        return $"Спикер: {SpeakerName}";
    }
}

ConcertTicket concert = new ConcertTicket(1, 10, 5, "Рок-группа");
PlayTicket play = new PlayTicket(2, 5, 12, "Гамлет");

ConferenceTicket conferenceIT = new ConferenceTicket(3, 2, 12, "IT-Конференция");
ConferenceTicket conferenceEngener = new ConferenceTicket(4, 4, 12, "Инженерная конференция");

concert.Genre = "Рок";
concert.SetEventInfo("Рок-фестиваль", 1500, "Главный зал", DateTime.Now.AddDays(10));

play.Director = "Иванов";
play.DurationMinutes = 150;
play.SetEventInfo("Гамлет", 800, "Малый зал", DateTime.Now.AddDays(5));

conferenceIT.SpeakerName = "Петров";
conferenceIT.SetEventInfo("IT-Конференция", 2000, "Конференц-зал", DateTime.Now.AddDays(20));

conferenceEngener.SpeakerName = "Сидоров";
conferenceEngener.SetEventInfo("Инженерная конференция", 1800, "Зал Б", DateTime.Now.AddDays(30));

concert.PrintTicketInfo();
Console.WriteLine(concert.GetTicketDetails());
Console.WriteLine(concert.GetArtistGenre());
Console.WriteLine(concert.IsRockConcert());
Console.WriteLine(concert.GetEventSummary());
Console.WriteLine(concert.IsExpensive());
Console.WriteLine(concert.DaysUntilEvent());
concert.ReserveSeat();
Console.WriteLine(concert.GetTicketDetails());

Console.WriteLine();

play.PrintTicketInfo();
Console.WriteLine(play.GetTicketDetails());
Console.WriteLine(play.GetPlayInfo());
Console.WriteLine(play.IsLongPlay());
Console.WriteLine(play.GetEventSummary());
play.ReserveSeat();
Console.WriteLine(play.GetTicketDetails());

Console.WriteLine();

conferenceIT.PrintTicketInfo();
Console.WriteLine(conferenceIT.GetTicketDetails());
Console.WriteLine(conferenceIT.HasSpeaker());
Console.WriteLine(conferenceIT.GetSpeakerInfo());
Console.WriteLine(conferenceIT.GenerateReport());
Console.WriteLine("Введите желаемое место для конференции:");

Console.SetIn(new StringReader("12"));
conferenceIT.ReserveSeat();

Console.WriteLine();

conferenceIT.PrintTicketInfo();

Console.WriteLine(conferenceEngener.GetTicketDetails());
Console.WriteLine(conferenceEngener.GenerateReport());
Console.WriteLine("Введите желаемое место для конференции:");

Console.SetIn(new StringReader("12"));
conferenceEngener.ReserveSeat();

Console.WriteLine();

Console.WriteLine($"Места в конференциях {conferenceIT.ConferenceName} и {conferenceEngener.ConferenceName} {(conferenceIT.EqualSeat(conferenceEngener) ? "совпадают" : "не совпадают")}");
```
